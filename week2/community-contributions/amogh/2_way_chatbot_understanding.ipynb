```python
import os
from IPython.display import display,Markdown
```

```python
def call_gpt(i):
    messages = [{"role":"system","content":"gpt_system_msg"}]
    for g,c in zip(gpt_msg,claude_msg):
        messages.append({"role":"assistant","content":g})
        messages.append({"role":"user","content":c})
    print(f"GPT {i}: {messages}\n")
    return f"gpt_{i}"
```

```python
def call_claude(i):
    messages = [{"role":"system","content":"claude_system_msg"}]
    for g,c in zip(gpt_msg,claude_msg):
        messages.append({"role":"user","content":g})
        messages.append({"role":"assistant","content":c})
    messages.append({"role":"user","content":gpt_msg[-1]})
    print(f"Claude {i}: {messages}\n")
    return f"claude_{i}"
```

```python
gpt_msg = ["gpt_1"]
claude_msg = ["claude_1"]

gpt_counter=1
claude_counter=1

display(Markdown(f"### GPT:\n{gpt_msg[0]}\n"))
display(Markdown(f"### Claude:\n{claude_msg[0]}\n"))

for i in range(5):

    display(Markdown(f"## Round {i}"))

    gpt_counter+=1
    gpt_next=call_gpt(gpt_counter)
    display(Markdown(f"### GPT:\n{gpt_next}\n"))
    gpt_msg.append(gpt_next)

    claude_counter+=1
    claude_next=call_claude(claude_counter)
    display(Markdown(f"### Claude:\n{claude_next}\n"))
    claude_msg.append(claude_next)
```
