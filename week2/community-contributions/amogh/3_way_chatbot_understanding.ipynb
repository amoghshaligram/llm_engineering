```python
import os
from IPython.display import display,Markdown
```

```python
def call_gpt(i):
    messages = [{"role":"system","content":"gpt_system_msg"}]
    for g,c,o in zip(gpt_msg,claude_msg,ollama_msg):
        messages.append({"role":"assistant","content":f"GPT: {g}"})
        messages.append({"role":"user","content":f"Claude: {c}\nOllama: {o}"})
    print(f"GPT {i}: {messages}\n")
    return f"gpt_{i}"
```

```python
def call_claude(i):
    messages = [{"role":"system","content":"claude_system_msg"}]
    for g,c,o in zip(gpt_msg,claude_msg,ollama_msg):
        messages.append({"role":"user","content":f"GPT: {g}\nOllama: {o}"})
        messages.append({"role":"assistant","content":f"Claude: {c}"})
    messages.append({"role":"user","content":f"GPT: {gpt_msg[-1]}"})
    print(f"Claude {i}: {messages}\n")
    return f"claude_{i}"
```

```python
def call_ollama(i):
    messages = [{"role":"system","content":"ollama_system_msg"}]
    for g,c,o in zip(gpt_msg,claude_msg,ollama_msg):
        messages.append({"role":"user","content":f"GPT: {g}\nClaude: {c}"})
        messages.append({"role":"assistant","content":f"Ollama: {o}"})
    messages.append({"role":"user","content":f"GPT: {gpt_msg[-1]}\nClaude: {claude_msg[-1]}"})
    print(f"Ollama {i}: {messages}\n")
    return f"ollama_{i}"
```

```python
gpt_msg = ["gpt_1"]
claude_msg = ["claude_1"]
ollama_msg = ["ollama_1"]

gpt_counter=1
claude_counter=1
ollama_counter=1

display(Markdown(f"## Round 0"))
display(Markdown(f"### GPT:\n{gpt_msg[0]}\n"))
display(Markdown(f"### Claude:\n{claude_msg[0]}\n"))
display(Markdown(f"### Ollama:\n{ollama_msg[0]}\n"))

for i in range(5):

    display(Markdown(f"## Round {i+1}"))

    gpt_counter+=1
    gpt_next=call_gpt(gpt_counter)
    display(Markdown(f"### GPT:\n{gpt_next}\n"))
    gpt_msg.append(gpt_next)

    claude_counter+=1
    claude_next=call_claude(claude_counter)
    display(Markdown(f"### Claude:\n{claude_next}\n"))
    claude_msg.append(claude_next)

    ollama_counter+=1
    ollama_next=call_ollama(ollama_counter)
    display(Markdown(f"### Ollama:\n{ollama_next}\n"))
    ollama_msg.append(ollama_next)

    print("-" * 100)
```
