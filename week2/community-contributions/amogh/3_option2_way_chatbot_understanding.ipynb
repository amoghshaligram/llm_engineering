```python
from IPython.display import display,Markdown
```

```python
gpt_system_msg = "gpt_system_msg"
claude_system_msg = "claude_system_msg"
ollama_system_msg = "ollama_system_msg"
```

```python
def call_llm(i,c,llm,s):
    messages = [
        {"role":"system","content":s},
        {"role":"user","content":c}
    ]
    display(Markdown(f"{llm} {i}: {messages}\n"))
    return f"{llm}: {llm}_{i}"
```

```python
def llm_user_prompt(llm,c,u1,u2):
    user_prompt = f"""
        You are {llm}, in conversation with {u1} and {u2}.
        The conversation so far is as follows:
        {c}
        Now with this, respond with what you would like to say next, as {llm}.
        """
    return user_prompt
```

```python
conversation = ""

conversation += f"gpt: gpt_1\n"
conversation += f"claude: claude_1\n"
conversation += f"ollama: ollama_1\n"

gpt_counter=1
claude_counter=1
ollama_counter=1

display(Markdown(f"## Round 0"))
display(Markdown(f"{conversation}\n"))
print("-" * 75)

for i in range(2):

    display(Markdown(f"## Round {i+1}\n"))

    gpt_counter+=1
    gpt_user_msg = f"{conversation}"
    gpt_next=call_llm(gpt_counter,llm_user_prompt("gpt",conversation,"claude","ollama"),"gpt",gpt_system_msg)
    display(Markdown(f"### GPT:\n{gpt_next}\n"))
    conversation += f"{gpt_next}\n"

    claude_counter+=1
    claude_user_msg = f"{conversation}"
    claude_next=call_llm(claude_counter,llm_user_prompt("claude",conversation,"gpt","ollama"),"claude",claude_system_msg)
    display(Markdown(f"### Claude:\n{claude_next}\n"))
    conversation += f"{claude_next}\n"

    ollama_counter+=1
    ollama_user_msg = f"{conversation}"
    ollama_next=call_llm(ollama_counter,llm_user_prompt("ollama",conversation,"gpt","claude"),"ollama",ollama_system_msg)
    display(Markdown(f"### Ollama:\n{ollama_next}\n"))
    conversation += f"{ollama_next}\n"

    print("-" * 100)
```
